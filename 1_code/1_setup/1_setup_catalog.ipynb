## Create schema for company using architeture medallion (bronze, silver  and gold)

```python
%sql

--schema parent
CREATE SCHEMA IF NOT EXISTS  atlone_enterprise.bronze_parent;
CREATE SCHEMA IF NOT EXISTS atlone_enterprise.silver_parent;

--schema child
CREATE SCHEMA IF NOT EXISTS  atlone_enterprise.bronze_child;
CREATE SCHEMA IF NOT EXISTS atlone_enterprise.silver_child;

--schema unified
CREATE SCHEMA IF NOT EXISTS atlone_enterprise.gold;
```
